```python
import requests
from bs4 import BeautifulSoup as bs

cabeceras = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.11'}

url = "https://farmaciascarrascosa.com/nutricion-y-dietetica/pastillas-adelgazar/bloqueador-grasas/"
page = requests.get(url, headers=cabeceras)

if page.status_code != 200:
    print('Error en la petición. Código:', page.status_code)


else:
    soup = bs(page.content, 'lxml')
    # print(soup.prettify())
    products = soup.find("ul", class_="products")
    # print(products.prettify())
    products = products.find_all("li")
    for product in products:
        name= product.find("p",class_="woocommerce-loop-product__title").text.strip()
        price= product.find("bdi").text.strip()
        url= product.find("a")["href"]
        print(f"Producto: {name}, Precio: {price}, Url: {url}")


```
